```python
import pandas as pd

df = pd.read_json("../data/data.json")

print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

df.head()
```

```python
print("Missing values:")
print(df.isna().sum())

print("\nTop origin countries:")
print(df["origin_country"].value_counts().head(10))

print("\nTop receiving countries:")
print(df["receiving_country"].value_counts().head(10))

print("\nTransit time summary:")
print(df["time"].describe())

print("\nDistance summary:")
print(df["distance"].describe())
```

```python
print("Top origin countries:")
print(df["origin_country"].value_counts().head(10))

print("\nTop receiving countries:")
print(df["receiving_country"].value_counts().head(10))

print("\nTransit time summary:")
print(df["time"].describe())

print("\nDistance summary:")
print(df["distance"].describe())
```

```python
import matplotlib.pyplot as plt

top_receiving = df["receiving_country"].value_counts().head(10)

plt.figure(figsize=(10, 6))
top_receiving.plot(kind="bar")
plt.title("Top 10 Receiving Countries")
plt.xlabel("Receiving Country")
plt.ylabel("Number of Postcards")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
top_origin = df["origin_country"].value_counts().head(10)

plt.figure(figsize=(10, 6))
top_origin.plot(kind="bar")
plt.title("Top 10 Origin Countries")
plt.xlabel("Origin Country")
plt.ylabel("Number of Postcards")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df["time"], bins=30)
plt.title("Distribution of Transit Time")
plt.xlabel("Transit Time (days)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df["distance"], bins=30)
plt.title("Distribution of Distance")
plt.xlabel("Distance (km)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df["distance"], df["time"], alpha=0.5)
plt.title("Distance vs Transit Time")
plt.xlabel("Distance (km)")
plt.ylabel("Transit Time (days)")
plt.tight_layout()
plt.show()
```
