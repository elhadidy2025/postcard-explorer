```python
import pandas as pd

df = pd.read_json("../data/data.json")

display_columns = [
    "id",
    "name",
    "origin_country",
    "receiving_country",
    "date_sent",
    "date_received",
    "distance",
    "time"
]

m1_df = df[display_columns].copy()

m1_df.head()
```

```python
filtered_df = m1_df[m1_df["receiving_country"] == "Germany"]

filtered_df.head(10)
```

```python
filtered_df = m1_df[
    (m1_df["receiving_country"] == "Germany") &
    (m1_df["time"] > 100)
]

filtered_df.head(10)
```

```python
sorted_df = filtered_df.sort_values(by="time", ascending=False)

sorted_df.head(10)
```

```python
print("Number of matching postcards:", len(sorted_df))
print("Average transit time:", round(sorted_df["time"].mean(), 1))
print("Average distance:", round(sorted_df["distance"].mean(), 1))
```
